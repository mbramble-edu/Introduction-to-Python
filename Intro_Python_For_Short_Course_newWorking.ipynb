### The `range()` function

`range()` produces a sequence of integers, often used to repeat an action in a `for` loop.

| Syntax | Meaning | Example | Numbers produced |
|---|---|---|---|
| `range(stop)` | Start at 0 | `range(5)` | 0, 1, 2, 3, 4 |
| `range(start, stop)` | Choose the starting number | `range(2, 6)` | 2, 3, 4, 5 |
| `range(start, stop, step)` | Choose the increment | `range(1, 10, 2)` | 1, 3, 5, 7, 9 |

**The stopping value is always excluded.** The default starting value is 0, and the default step is 1.

Use a negative step to count backward: `range(5, 0, -1)` produces 5, 4, 3, 2, 1.

`range()` creates a range object rather than a list. Use `list(range(5))` to see its numbers as a list.

### More advanced: Combining `break` and `continue` in loops
-And introducing the jupyterlab debugger

Search through a list of usernames, skip empty entries,
and stop as soon as we find a match.

If this is easy, move on to the next GC content example and find the problems.

````python
usernames = ["alice", "", "bob", "carol", "", "dave"]
target = "carol"

for user in usernames:
    if user == "":
        continue            # skip blank entries, keep looping
    print(f"Checking user: {user}")
    if user == target:
        print(f"Found {target}! Stopping search.")
        break
    else:
        # This 'else' belongs to the FOR loop -- it runs only if the loop
        # finished WITHOUT hitting 'break'. Handy for "not found" cases.
        print("User not found.")
print("Done")
````

## 11. Functions

A **function** is a reusable block of code. You define one with `def`, and it can take
**parameters** (inputs) and give back a **return value** (output).

### Why use functions?
- Avoid repeating yourself
- Break a big problem into small, named, testable pieces


### Functions: reusable blocks of code

A **function** groups instructions under a name so you can run them whenever you need them. It can receive **inputs**, perform an operation, and **return an output**.

$$
\boxed{\text{Input: } 5}
\;\longrightarrow\;
\boxed{\text{5*2}}
\;\longrightarrow\;
\boxed{\text{Output: } 10}
$$

```python
def double(number):
    """Return twice the input number."""
    result = number * 2
    return result

answer = double(5)
print(answer)           # 10
```

- **`def`** begins the function definition.
- **`double`** is the function’s name.
- **`number`** is a **parameter**: a name for the input inside the function.
- The indented lines form the function’s **body**.
- **`return`** sends a value back to the code that called the function.

Defining a function does **not** run its body. Calling `double(5)` runs it, with the **argument** `5` assigned to the parameter `number`.

### A more complex loop with a function call : Using the debugger to follow complex code


````python
sequences = ["ATGC", "ATNGC", "GGXCTA", "ATATAT", "I'm invalid"]
nucs = "ATGCN"
def get_percent(gc_count_arg, base_count_arg):
    gc = 100 * gc_count_arg / base_count_arg
    return(gc)

percent_dict = {}
for sequence in sequences:
    gc_count = 0
    base_count = 0
    invalid = False
    for base in sequence:
        if not base in nucs:
            invalid = True
        base_count += 1
        if base in "GC":
            gc_count += 1
    if invalid:
        continue  # Skip the calculation; move to the next sequence
    if base_count == 0:
        print(f"{sequence}: no known bases")
        continue
    gc_percent = get_percent(gc_count, base_count)
    percent_dict[sequence] = gc_percent
    print(f"{sequence}: {gc_percent:.1f}% GC")
percent_dict
````

````python
sequences = ["ATGCGC", "ATNNGC", "GGXXCTA", "ATATAT", "I'm invalid"]
nucs = "ATGCN"


def get_percent(gc_count_arg, base_count_arg):
    gc = 100 * gc_count_arg / base_count_arg
    return(gc)

def run_sequences(nucs_arg, sequences_arg):
    for sequence in sequences_arg:
        gc_count = 0
        offending_bases = []
        invalid = False
        for base in sequence:
            if not base in nucs_arg:
                offending_bases.append(base)
                invalid = True
            if base in "GC":
                gc_count += 1
        base_count = len(sequence)
        gc_percent = get_percent(gc_count, base_count)
        print(f"{sequence}: {gc_percent:.1f}% GC")
run_sequences(nucs, sequences)
````

````python
#The following should modify the values in a dictionary of read counts.

def add_reads(current_count, new_reads):
    updated_count = current_count + new_reads
    return updated_count


def update_counts(sequence_counts, new_reads):
    """Add new_reads to the count for every sequence."""
    for sequence, count in sequence_counts.items():
        updated_count = add_reads(count, new_reads)
        sequence_counts[sequence] = updated_count
        return sequence_counts


read_counts = {
    "ATGC": 12,
    "GGTA": 8,
    "CCAT": 20,
    "TATA": 0
}

result = update_counts(read_counts, 5)
print(result)

#expected output: all counts should be altered in the new dictionary
#{'ATGC': 17, 'GGTA': 13, 'CCAT': 25, 'TATA': 5}
````

````python
#The following code modifies prices in a dictionary of item prices

def apply_discount(price, rate):
    """Return the price after applying a discount rate (e.g. 0.10 = 10% off)."""
    return round(price * (1 - rate), 2)


def discount_all(prices, rate):
    """Loop over a dictionary of item prices and discount each one."""
    for item, price in prices.items():
        price = apply_discount(price, rate)
    return prices


store_prices = {
    "notebook": 4.50,
    "backpack": 32.00,
    "pencil case": 6.25,
    "calculator": 18.99,
}

print("Before:", store_prices)
discounted = discount_all(store_prices, 0.10)
print("After: ", discounted)

# Expected output:
# Before: {'notebook': 4.5, 'backpack': 32.0, 'pencil case': 6.25, 'calculator': 18.99}
# After:  {'notebook': 4.05, 'backpack': 28.8, 'pencil case': 5.62, 'calculator': 17.09}

````
